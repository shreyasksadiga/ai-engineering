### Step 1: Import Libraries and API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing")
```

### Step 2: Searching the web

```python
client = OpenAI()

completion = client.chat.completions.create(
    model="gpt-4o-mini-search-preview",
    messages=[
        {"role": "user", "content": "What are the latest AI confrerences happening in Dubai?"}
    ],
    max_tokens=500
)
```

```python
search_results = completion.choices[0].message.content

display(Markdown(search_results))
```

### Step 3: Summarize and write a social media post

```python
system_message = """
You are a helpful assistant whos job is to accurately summarize long series of 
news articles provided to you by the user and write up a compelling and easy to 
digest social media post.
"""

prompt = f"""
The following is a series of news articles on the field of AI. Please summarize
the articles and write a compelling social media post for LinkedIn highlighting
key points about each article.

The post should be short and concise and should not be more than 100 words.

{search_results}

"""

completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
    ],
    max_tokens=500
)

chat_response = completion.choices[0].message.content

print(chat_response)
```

 **Creating my own post for linkdin**

```python
import os
from openai import OpenAI
from IPython.display import Markdown, display
from dotenv import load_dotenv


load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
if OPENAI_API_KEY is None : 
    raise Exception("API key is missing")
```

```python
client = OpenAI()
chat_message = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[
        {
            "role" : "user",
            "content" : "give me the latest update on the AI in 2026"
        }
    ],
    max_tokens=300
)
response = chat_message.choices[0].message.content
display(Markdown(response));
```

```python
system_message = 'you are the AI Assistent, you have good knowledge on how to post something in the social media'

user_prompt = f"""
I have details on the lastest inovation about the AI,
here is the details : {response}
Need to post this in the instagram give me the proper post.
"""
```

```python
post_creation = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[
        {
            'role' : "system",
            'content' : system_message
        },
        {
            "role" : "user",
            "content" : user_prompt
        }
    ],
    max_tokens=500
)
display(Markdown(post_creation.choices[0].message.content))
```
